```python
def reorder_list(similarity_list, target_word):
    # 'target_word'와 관련된 항목을 찾음
    target_item = next((item for item in similarity_list if item[0] == target_word), None)

    # 'target_word'가 목록에 있으면 최상위로 이동
    if target_item:
        similarity_list.remove(target_item)
        similarity_list.insert(0, target_item)

    return similarity_list

def find_related_menus(model, ingredient, recipe_dataset, topn=10):
    # Word2Vec 모델을 사용하여 입력된 재료와 유사한 단어 찾기
    similar_words = model.wv.most_similar(ingredient, topn=topn)

    # 유사한 단어 목록을 재정렬
    reordered_words = reorder_list(similar_words, ingredient)

    # 유사한 단어와 연관된 메뉴명 찾기
    for word, similarity in reordered_words:
        similar_menus = recipe_dataset[recipe_dataset['CKG_MTRL_CN'].str.contains(word, case=False)]
        print(f"Similar Word: {word}, Similarity: {similarity}")
        print("Associated Menus:")
        print(similar_menus[['CKG_NM']])
        print("\n")

def find_ingredients_by_menu_name(menu_name, recipe_dataset):
    # 메뉴명을 기반으로 재료 찾기
    matched_menus = recipe_dataset[recipe_dataset['CKG_NM'].str.contains(menu_name, case=False)]
    if not matched_menus.empty:
        print(f"Ingredients for {menu_name}:")
        for index, row in matched_menus.iterrows():
            print(row['CKG_MTRL_CN'])
    else:
        print(f"No menus found for {menu_name}")
```

```python
def find_related_menus(model, ingredient, recipe_dataset, topn=10):
    # Word2Vec 모델을 사용하여 입력된 재료와 유사한 단어 찾기
    similar_words = model.wv.most_similar(ingredient, topn=topn)

    # '갈치'가 유사 단어 목록에 없으면 목록에 추가
    if not any(word == ingredient for word, _ in similar_words):
        similar_words.append((ingredient, 1.0))

    # 유사한 단어 목록을 재정렬
    reordered_words = reorder_list(similar_words, ingredient)

    # 유사한 단어와 연관된 메뉴명 찾기
    for word, similarity in reordered_words:
        similar_menus = recipe_dataset[recipe_dataset['CKG_MTRL_CN'].str.contains(word, case=False)]
        print(f"Similar Word: {word}, Similarity: {similarity}")
        print("Associated Menus:")
        print(similar_menus[['CKG_NM']].drop_duplicates())
        print("\n")
```
